# BER 2/2: cross-encoder rescoring (GPU)

**Setup**
- **Add Input** → **Notebooks** → add the finished notebook **BER 1/2** (its output contains `ce_pkg/`).
- **Settings**: Accelerator = **GPU T4 x2** or **P100**, **Internet = On**.
- **Save Version → Save & Run All**. About 1 hour for `distilbert`, 2–2.5 hours for `xlm-roberta-base`.
- Result: `final/matching_results.tsv` (upload this), `final/matching_results_stricter.tsv`, `final/candidate_pairs.tsv`, `final/ce_report.json`.

```python
import glob, os, subprocess, sys, json
pk = glob.glob('/kaggle/input/**/ce_pkg/ce_train.parquet', recursive=True)
assert pk, 'add the BER 1/2 notebook output as input'
PKG = os.path.dirname(pk[0]); print(PKG, sorted(os.listdir(PKG)))
MODEL = 'distilbert-base-multilingual-cased'   # Apache-2.0. Alternative: 'xlm-roberta-base' (MIT, slower, usually stronger)
EPOCHS = 1
```

```python
p = subprocess.run([sys.executable, f'{PKG}/ce_train.py', '--pkg', PKG, '--out', '/kaggle/working/final',
                    '--model', MODEL, '--epochs', str(EPOCHS), '--workers', '4'])
assert p.returncode == 0
print(json.dumps(json.load(open('/kaggle/working/final/ce_report.json')), indent=1))
!ls -la /kaggle/working/final
```
